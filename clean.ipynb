Opening the data file and looking how to clean the data 

checking that the cleaning method is functional

Saving the clean data as /data/census_clean.csv

```python
import pandas as pd

data = pd.read_csv("./data/census.csv")
for column in data.select_dtypes(include="object").columns:
    print(f"\n{column}:")
    print([repr(x) for x in data[column].unique()[:10]])
```

```python
data.columns = data.columns.str.strip()
data = data.apply(lambda column: column.str.strip() if column.dtype == "object" else column)
for column in data.select_dtypes(include="object").columns:
    print(f"\n{column}:")
    print([repr(x) for x in data[column].unique()[:10]])
```

```python
data.to_csv("./data/census_clean.csv")
```
